### Loading user data from the CSV file

```python
import pandas as pd
df = pd.read_csv("users.csv")
display(df)
```

### Initializing the SQLite database and creating the 'users' table

```python
import sqlite3

connection = sqlite3.connect("users.db")
cursor = connection.cursor()

cursor.execute("""
    CREATE TABLE IF NOT EXISTS users (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT NOT NULL,
        email TEXT NOT NULL
    )
""")

connection.commit()

print("Database created successfully!")
```

### Iterating through the DataFrame and inserting each record into the database

```python
for _, row in df.iterrows():

    cursor.execute("""
        INSERT INTO users (name, email)
        VALUES (?, ?)
    """, (
        row["name"],
        row["email"]
    ))

connection.commit()

print("CSV data inserted successfully!")
```

### Querying and displaying all records stored in the 'users' table

```python
result = pd.read_sql_query(
    "SELECT * FROM users",
    connection
)

display(result)
```
